```python
#Animal(Dog)

class Animal:
    def sound(self):
        print("Animal Sound")

class Dog(Animal):
    pass

Dog().sound()
```

```python
#Person(Student)

class Person:
    def show(self):
        print("Person")

class Student(Person):
    pass

Student().show()
```

```python
#Vehicle(Bike)

class Vehicle:
    def start(self):
        print("Vehicle Started")

class Bike(Vehicle):
    pass

Bike().start()
```

```python
#Shape(Circle)

class Shape:
    def draw(self):
        print("Drawing Shape")

class Circle(Shape):
    pass

Circle().draw()
```

```python
#Employee(Manager)

class Employee:
    def work(self):
        print("Working")

class Manager(Employee):
    pass

Manager().work()
```
