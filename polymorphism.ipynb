```python
#Dog & Cat

class Dog:
    def sound(self):
        print("Bark")

class Cat:
    def sound(self):
        print("Meow")

for i in (Dog(), Cat()):
    i.sound()
```

```python
#Car & Bike

class Car:
    def move(self):
        print("Car Moves")

class Bike:
    def move(self):
        print("Bike Moves")

for i in (Car(), Bike()):
    i.move()
```

```python
#Teacher & Student

class Teacher:
    def work(self):
        print("Teaching")

class Student:
    def work(self):
        print("Learning")

for i in (Teacher(), Student()):
    i.work()
```

```python
#Bird & Fish

class Bird:
    def action(self):
        print("Flying")

class Fish:
    def action(self):
        print("Swimming")

for i in (Bird(), Fish()):
    i.action()
```

```python
#Lion & Cow

class Lion:
    def sound(self):
        print("Roar")

class Cow:
    def sound(self):
        print("Moo")

for i in (Lion(), Cow()):
    i.sound()
```
