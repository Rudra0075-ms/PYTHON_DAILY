```python
#Student Details

class Student:
    def __init__(self):
        self.__marks = 90
    def show(self):
        print(self.__marks)

s = Student()
s.show()
```

```python
#Bank Account

class Bank:
    def __init__(self):
        self.__bal = 5000
    def balance(self):
        print(self.__bal)

b = Bank()
b.balance()
```

```python
#Employee Salary

class Employee:
    def __init__(self):
        self.__salary = 30000
    def show(self):
        print(self.__salary)

Employee().show()
```

```python
#Mobile Price

class Mobile:
    def __init__(self):
        self.__price = 15000
    def display(self):
        print(self.__price)

Mobile().display()
```

```python
#Car Speed

class Car:
    def __init__(self):
        self.__speed = 120
    def show(self):
        print(self.__speed)

Car().show()
```
